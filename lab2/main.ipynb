## 1. Thay đổi độ sáng (Brightness)

Độ sáng được thay đổi bằng cách cộng hoặc trừ một giá trị cố định vào tất cả các điểm ảnh.

**Công thức:**

\[
g(x,y)=f(x,y)+\beta
\]

- β > 0 : Tăng độ sáng
- β < 0 : Giảm độ sáng

```python
image = cv2.imread("datatest/test.jpg")

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
#Tăng độ sáng của ảnh
brightness = 50

bright_image = cv2.convertScaleAbs(
    image,
    alpha=1,
    beta=brightness
)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(bright_image)
plt.title("Brightness +50")
plt.axis("off")

plt.show()
#Giảm độ sáng của ảnh
dark_image = cv2.convertScaleAbs(
    image,
    alpha=1,
    beta=-50
)

plt.figure(figsize=(6,6))
plt.imshow(dark_image)
plt.title("Brightness -50")
plt.axis("off")
plt.show()
```

## 2. Thay đổi độ tương phản

Độ tương phản được thay đổi bằng cách nhân mỗi điểm ảnh với hệ số α.

Công thức:

g(x,y)=αf(x,y)

α > 1: tăng tương phản

0 < α < 1: giảm tương phản

```python
#Tăng độ tương phản
high_contrast = cv2.convertScaleAbs(
    image,
    alpha=1.5,
    beta=0
)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(high_contrast)
plt.title("Contrast x1.5")
plt.axis("off")

plt.show()
#Giảm độ tương phản
low_contrast = cv2.convertScaleAbs(
    image,
    alpha=0.5,
    beta=0
)

plt.figure(figsize=(6,6))
plt.imshow(low_contrast)
plt.title("Contrast x0.5")
plt.axis("off")
plt.show()
```

## 3.Biến đổi âm bản

Thực hiện biến đổi âm bản (Negative Transformation) cho một hình ảnh bằng thư viện  Matplotlib trong Python với công thức

s = L - 1 - r

Trong đó:

r: Giá trị của điểm ảnh gốc.

s: Giá trị của điểm ảnh sau khi biến đổi (âm bản).

L: Số mức xám/màu của ảnh

Với ảnh 8-bit chuẩn (RGB), L = 256, do đó L - 1 = 255.

Sẽ cho ra kết quả 

Bên trái: Ảnh gốc hệ màu RGB.

Bên phải: Ảnh âm bản đã biến đổi.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# Đường dẫn đến file ảnh trong folder datatest
image_path = os.path.join('..', 'datatest', 'test.jpg')


# Đọc ảnh
img = cv2.imread(image_path)

if img is None:
    print(f"Không tìm thấy ảnh tại: {image_path}. Bạn hãy kiểm tra lại tên file ảnh trong folder 'datatest' nhé!")
else:
    # Chuyển từ BGR sang RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # 3. Biến đổi âm bản: Đảo ngược các giá trị điểm ảnh
    negative_img = 255 - img_rgb

    # Hiển thị kết quả
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title('Ảnh gốc (RGB)')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(negative_img)
    plt.title('Phần 3: Ảnh âm bản')
    plt.axis('off')
    plt.show()
```

### II. Làm sắc nét

Làm sắc nét: Tăng cường các cạnh trong ảnh bằng cách sử dụng các kernel thích hợp.

1. Kernel cơ bản: Nhân giá trị điểm ảnh trung tâm với hệ số lớn, trừ đi các điểm ảnh lân cận (trên, dưới, trái, phải) để khuếch đại sự khác biệt tại các cạnh, giữ nguyên vùng ảnh phẳng.
2. Kernel mạnh hơn: Trừ cả 8 điểm ảnh lân cận (bao gồm đường chéo), giúp cạnh rõ nét hơn nhiều nhưng cũng dễ khuếch đại nhiễu nếu ảnh gốc không sạch.
3. Unsharp masking: Làm mờ ảnh gốc bằng Gaussian blur để tách phần chi tiết cao tần (ảnh gốc trừ ảnh mờ), sau đó cộng ngược phần chi tiết đó vào ảnh gốc với hệ số điều chỉnh được, cho kết quả tự nhiên hơn.

### 1. Kernel cơ bản

Công thức kernel:

|  0 | -1 |  0 |
| -1 |  5 | -1 |
|  0 | -1 |  0 |

Với mỗi điểm ảnh, giá trị mới được tính:

pixel_mới = 5 × pixel_tâm - (pixel_trên + pixel_dưới + pixel_trái + pixel_phải)

Tổng các hệ số trong kernel = 1, giúp giữ nguyên độ sáng trung bình của ảnh, chỉ khuếch đại chênh lệch tại các cạnh.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def sharpen_basic(image):
    """Làm sắc nét ảnh bằng kernel cơ bản.

    Args:
        image: Ảnh đầu vào dạng NumPy array.

    Returns:
        Ảnh đã được làm sắc nét, dạng NumPy array.
    """
    kernel = np.array([
        [ 0, -1,  0],
        [-1,  5, -1],
        [ 0, -1,  0]
    ])
    sharpened = cv2.filter2D(image, -1, kernel)
    return sharpened


image = cv2.imread('../datatest/test.jpg')
sharpened_image = sharpen_basic(image)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Ảnh gốc')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(sharpened_image, cv2.COLOR_BGR2RGB))
plt.title('Kernel cơ bản')
plt.axis('off')
plt.show()
```

### 2. Kernel mạnh hơn

Công thức kernel:

| -1 | -1 | -1 |
| -1 |  9 | -1 |
| -1 | -1 | -1 |

Với mỗi điểm ảnh, giá trị mới được tính:

pixel_mới = 9 × pixel_tâm - tổng(8 pixel lân cận, kể cả đường chéo)

So với kernel cơ bản, kernel này trừ thêm 4 pixel đường chéo, tổng hệ số vẫn = 1 nhưng mức khuếch đại cạnh mạnh hơn nhiều.

```python
def sharpen_strong(image):
    """Làm sắc nét ảnh bằng kernel mạnh hơn.

    Args:
        image: Ảnh đầu vào dạng NumPy array.

    Returns:
        Ảnh đã được làm sắc nét, dạng NumPy array.
    """
    kernel = np.array([
        [-1, -1, -1],
        [-1,  9, -1],
        [-1, -1, -1]
    ])
    sharpened = cv2.filter2D(image, -1, kernel)
    return sharpened


strong_image = sharpen_strong(image)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Ảnh gốc')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(strong_image, cv2.COLOR_BGR2RGB))
plt.title('Kernel mạnh hơn')
plt.axis('off')
plt.show()
```

### 3. Unsharp masking

Công thức:

ảnh_mờ = GaussianBlur(ảnh_gốc, sigma)
ảnh_sắc_nét = ảnh_gốc + strength × (ảnh_gốc - ảnh_mờ)

Trong đó:
- sigma: độ lệch chuẩn của Gaussian blur, quyết định vùng ảnh hưởng
- strength: hệ số khuếch đại phần chi tiết (ảnh_gốc - ảnh_mờ) được cộng thêm vào ảnh gốc

```python
def sharpen_unsharp_mask(image, sigma=3, strength=1.5):
    """Làm sắc nét ảnh bằng kỹ thuật unsharp masking.

    Args:
        image: Ảnh đầu vào dạng NumPy array.
        sigma: Độ lệch chuẩn của Gaussian blur.
        strength: Hệ số khuếch đại phần chi tiết được cộng thêm.

    Returns:
        Ảnh đã được làm sắc nét, dạng NumPy array.
    """
    blurred = cv2.GaussianBlur(image, (0, 0), sigma)
    sharpened = cv2.addWeighted(image, 1 + strength, blurred, -strength, 0)
    return sharpened


unsharp_image = sharpen_unsharp_mask(image)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Ảnh gốc')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(unsharp_image, cv2.COLOR_BGR2RGB))
plt.title('Unsharp masking')
plt.axis('off')
plt.show()
```

## III.1 - Phát hiện cạnh: Sobel vs Prewitt
Trần Thị Minh Thùy-091306004639
III. Bài tập nâng cao - Mục 1: Phát hiện cạnh bằng kernel Sobel và Prewitt

Nội dung thực hiện:
1.Cài đặt thủ công phép tích chập 2D (convolution) bằng vòng lặp
2.Cài đặt phát hiện cạnh bằng toán tử Sobel (thủ công và dùng OpenCV `cv2.filter2D`)
3.Cài đặt phát hiện cạnh bằng toán tử Prewitt (thủ công và dùng OpenCV)
4.Ngưỡng hóa (thresholding) kết quả để thu được ảnh nhị phân biên cạnh
5.So sánh trực quan kết quả giữa cách tính thủ công và dùng thư viện OpenCV

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
# Tính tích chập 2D thủ công 
def convolve2d_manual(image, kernel):
    """
    Thực hiện phép toán tích chập (convolution) 2D thủ công bằng vòng lặp.
    """
    img_h, img_w = image.shape
    k_h, k_w = kernel.shape
    # Tính padding cần thiết để kernel trượt được tới cả pixel ở biên ảnh
    pad_h = k_h // 2
    pad_w = k_w // 2
    # Đệm viền ảnh bằng 0 (Zero Padding)
    padded_image = np.zeros((img_h + 2 * pad_h, img_w + 2 * pad_w), dtype=np.float32)
    padded_image[pad_h:pad_h + img_h, pad_w:pad_w + img_w] = image
    
    output = np.zeros((img_h, img_w), dtype=np.float32)
    # Lật kernel 180 độ — đúng định nghĩa toán học của convolution
    kernel_flipped = np.flip(np.flip(kernel, 0), 1)
    # Trượt kernel qua từng pixel của ảnh gốc
    for y in range(img_h):
        for x in range(img_w):
            neighborhood = padded_image[y:y + k_h, x:x + k_w]
            output[y, x] = np.sum(neighborhood * kernel_flipped)
            
    return output


#Toán tử SOBEL — phát hiện cạnh bằng 2 kernel Gx (cạnh dọc) và Gy (cạnh ngang)
def edge_detection_sobel(image, manual=True):
    """
    Phát hiện cạnh bằng toán tử Sobel.
    """
    # Kernel theo phương ngang — bắt cạnh dọc
    Kx = np.array([[-1, 0, 1],
                   [-2, 0, 2],
                   [-1, 0, 1]], dtype=np.float32)
    
    # Kernel theo phương dọc — bắt cạnh ngang
    Ky = np.array([[-1, -2, -1],
                   [ 0,  0,  0],
                   [ 1,  2,  1]], dtype=np.float32)
    
    if manual:
        # Dùng hàm tự viết ở trên
        Gx = convolve2d_manual(image, Kx)
        Gy = convolve2d_manual(image, Ky)
    else:
        # Đối chiếu với hàm có sẵn của OpenCV
        Gx = cv2.filter2D(image.astype(np.float32), -1, Kx)
        Gy = cv2.filter2D(image.astype(np.float32), -1, Ky)
    # Độ lớn gradient tổng hợp: G = sqrt(Gx^2 + Gy^2)        
    magnitude = np.sqrt(Gx**2 + Gy**2)
    # Chuẩn hóa về khoảng ảnh hợp lệ [0, 255]
    magnitude = np.clip(magnitude, 0, 255).astype(np.uint8)
    
    return magnitude, Gx, Gy
#Toán tử PREWITT(tương tự như Sobel nhưng kernel khác)
def edge_detection_prewitt(image, manual=True):
    """
    Phát hiện cạnh bằng toán tử Prewitt.
    """
    Kx = np.array([[-1, 0, 1],
                   [-1, 0, 1],
                   [-1, 0, 1]], dtype=np.float32)
    
    Ky = np.array([[-1, -1, -1],
                   [ 0,  0,  0],
                   [ 1,  1,  1]], dtype=np.float32)
    
    if manual:
        Gx = convolve2d_manual(image, Kx)
        Gy = convolve2d_manual(image, Ky)
    else:
        Gx = cv2.filter2D(image.astype(np.float32), -1, Kx)
        Gy = cv2.filter2D(image.astype(np.float32), -1, Ky)
        
    magnitude = np.sqrt(Gx**2 + Gy**2)
    magnitude = np.clip(magnitude, 0, 255).astype(np.uint8)
    
    return magnitude, Gx, Gy
#Ngưỡng hóa — chuyển ảnh gradient (nhiều mức xám) thành ảnh nhị phân (chỉ đen/trắng) để thấy rõ đường biên
def apply_threshold(magnitude, threshold_val=100):
    """
    Ngưỡng hóa (Thresholding) để thu được ảnh nhị phân biên cạnh rõ nét.
    """
    binary_edges = np.zeros_like(magnitude)
    # Pixel nào có gradient đủ lớn (>= threshold) thì coi là cạnh -> gán trắng
    binary_edges[magnitude >= threshold_val] = 255
    return binary_edges
```

```python
# Đọc ảnh xám từ dataset có sẵn
img_path = '../datatest/test.jpg'  
img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError(f"Không đọc được ảnh tại path: {img_path}")

print(f"Đã đọc ảnh kích thước {img.shape}")

# Sobel
sobel_manual, sobel_Gx_m, sobel_Gy_m = edge_detection_sobel(img, manual=True)
sobel_opencv, _, _ = edge_detection_sobel(img, manual=False)

# Prewitt
prewitt_manual, prewitt_Gx_m, prewitt_Gy_m = edge_detection_prewitt(img, manual=True)
prewitt_opencv, _, _ = edge_detection_prewitt(img, manual=False)

# Ngưỡng hóa
thresh = 100
sobel_thresh = apply_threshold(sobel_manual, thresh)
prewitt_thresh = apply_threshold(prewitt_manual, thresh)
```

**Nhận xét:**
Kết quả tính thủ công (`convolve2d_manual`) và dùng thư viện OpenCV (`cv2.filter2D`) cho ra ảnh gradient gần như giống hệt nhau ở cả Sobel và Prewitt, chứng tỏ hàm tích chập tự viết đã cài đặt đúng thuật toán.
Sobel Gx bắt rõ các cạnh dọc/xiên như vành nón và các sợi tóc, trong khi Sobel Gy phản ứng yếu hơn nhiều vì ảnh gốc chứa ít cạnh ngang rõ nét.
So sánh Sobel và Prewitt: hai kết quả khá tương đồng, nhưng Sobel cho đường biên sắc nét và ít nhiễu hơn một chút ở các chi tiết nhỏ (vùng tóc), do trọng số tâm của kernel Sobel lớn hơn (hệ số 2) giúp làm mượt nhiễu tốt hơn Prewitt.
Sau khi ngưỡng hóa (threshold=100), các đường viền chính của khuôn mặt và vành nón được giữ lại rõ ràng, tuy nhiên vùng tóc xoăn vẫn còn khá nhiều nhiễu dạng đốm nhỏ, có thể cải thiện bằng cách tăng ngưỡng hoặc áp dụng lọc làm mượt (Gaussian blur) trước khi phát hiện cạnh.

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python

#Dò cạnh theo chiều ngang (Sobel Horizontal) - Hiện rõ các đường nằm ngang
kernel_sobel_h = np.array([
        [-1, -2, -1],
        [ 0,  0,  0],
        [ 1,  2,  1]
    ], dtype=np.float32)
sobelh = cv2.filter2D(image_rgb, -1 , kernel_sobel_h)
plt.imshow(sobelh)        
```

```python
#Dò cạnh theo chiều dọc (Sobel Vertical)ng
kernel_sobel_v = np.array([
        [-1, 0, 1],
        [-2, 0, 2],
        [-1, 0, 1]
    ], dtype=np.float32)
sobelv = cv2.filter2D(image, -1 , kernel_sobel_v)
plt.imshow(sobelv)   
```

```python
#Kernel Laplacian (Dò cạnh mọi hướng - Toán tử vi phân bậc 2)
kernel_laplacian = np.array([
        [ 0,  1,  0],
        [ 1, -4,  1],
        [ 0,  1,  0]
    ], dtype=np.float32)
lab= cv2.filter2D(gray, cv2.CV_64F , kernel_laplacian)
labunit = np.uint8(np.abs(lab))
plt.imshow(labunit)   
```

```python
    # 4. Kernel Siêu Sắc Nét (Unsharp Masking / Excess Sharpen)
kernel_super_sharpen = np.array([
        [-1, -1, -1],
        [-1,  9, -1],
        [-1, -1, -1]
    ], dtype=np.float32)
super_sharpen=cv2.filter2D(image_rgb,-1,kernel_super_sharpen)
plt.imshow(super_sharpen)
```

```python
# 5. Kernel Dịch Chuyển Bóng  Tạo Hiệu Ứng Chuyển Động (Motion Blur ngang)
size = 9
kernel_motion = np.zeros((size, size), dtype=np.float32)
kernel_motion[int((size - 1) / 2), :] = 1.0 / size 
motion_blur=cv2.filter2D(image_rgb,-1,kernel_motion)
plt.imshow(motion_blur)
```

## So Sánh các loại lọc, so sánh hiệu quả của các loại lọc trên cùng 1 hình ảnh 

```python

# Lọc 1: Average Blur (Làm mờ trung bình)
blur_avg = cv2.blur(image_rgb, (5, 5))

    # Lọc 2: Gaussian Blur (Làm mờ mịn Gauss)
blur_gaussian = cv2.GaussianBlur(image_rgb, (5, 5), 0)

    # Lọc 3: Median Blur (Làm mờ trung vị - Giảm nhiễu tốt)
blur_median = cv2.medianBlur(image_rgb, 5)

    # Lọc 4: Sobel Filter 
sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
sobel_combined = cv2.magnitude(sobel_x, sobel_y)

    # Lọc 5: Laplacian Filter (Dò viền đa hướng)
laplacian = cv2.Laplacian(gray, cv2.CV_64F)
```

```python
import matplotlib.pyplot as plt

titles = [
    "1. Average Blur",
    "2. Gaussian Blur", 
    "3. Median Blur"
]

images = [
    blur_avg,
    blur_gaussian,
    blur_median
]

plt.figure(figsize=(15, 5))

for i in range(len(images)):
    plt.subplot(1, 3, i + 1)
    plt.imshow(images[i])
    plt.title(titles[i], fontsize=12, fontweight='bold')
    plt.axis('off')

plt.tight_layout()
plt.show()
```

**1. Nhóm lọc làm mờ (Smooth / Blur Filters):**
Average Blur (Lọc Trung Bình):

- Đặc điểm: Bằng phẳng hóa các giá trị pixel trong cửa sổ $5 \times 5$.

- Nhận xét: Làm mờ toàn bộ bức ảnh đều nhau. Các chi tiết nét nhỏ (như sợi lông trên mũ, tóc) bị nhòe đi đáng kể.

Gaussian Blur (Lọc Gauss):

- Đặc điểm: Sử dụng phân bố chuẩn Gauss (trọng số cao ở tâm và giảm dần ra mép).

- Nhận xét: Ảnh mờ mịn và tự nhiên hơn so với Average Blur. Giữ lại được cấu trúc tổng thể mượt mà hơn, hạn chế hiện tượng bị nhòe hộp (box effect).
Median Blur (Lọc Trung Vị):

- Đặc điểm: Lấy giá trị trung vị của vùng lân cận.

- Nhận xét: Giữ biên/cạnh (edges) của vật thể nét hơn hẳn so với Average và Gaussian Blur. Đây là bộ lọc tối ưu nhất để loại bỏ nhiễu muối tiêu (salt-and-pepper noise) mà vẫn giữ được độ rõ của đường nét.

```python
plt.subplot(1, 2, 1)
plt.imshow(sobel_combined, cmap='gray')
plt.title("4. Sobel Filter", fontsize=12, fontweight='bold')
plt.axis('off')

plt.subplot(1, 2, 2)
laplacian_clean = np.uint8(np.abs(laplacian)) 
plt.imshow(laplacian_clean, cmap='gray')
plt.title("5. Laplacian Filter", fontsize=12, fontweight='bold')
plt.axis('off')

plt.tight_layout()
plt.show()
```

**Nhóm lọc phát hiện cạnh / Dò viền (Edge Detection Filters)**

Sobel Filter:

- Đặc điểm: Toán tử đạo hàm bậc nhất (kết hợp cả phương ngang và phương dọc).

- Nhận xét: Làm nổi bật rõ ràng các đường biên chính của vật thể (vành mũ, đường nét khuôn mặt, vai) trên nền đen. Bị ít nhiễu nền hơn do có tích hợp thành phần làm mờ phẳng.

Laplacian Filter:

- Đặc điểm: Toán tử đạo hàm bậc hai (phát hiện cạnh đa hướng).

- Nhận xét: Nhạy cảm hơn với sự thay đổi độ sáng đột ngột, bắt được cả các chi tiết cạnh nhỏ và mảnh.
