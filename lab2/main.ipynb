## 1. Thay đổi độ sáng (Brightness)

Độ sáng được thay đổi bằng cách cộng hoặc trừ một giá trị cố định vào tất cả các điểm ảnh.

**Công thức:**

\[
g(x,y)=f(x,y)+\beta
\]

- β > 0 : Tăng độ sáng
- β < 0 : Giảm độ sáng

```python
image = cv2.imread("datatest/test.jpg")

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
#Tăng độ sáng của ảnh
brightness = 50

bright_image = cv2.convertScaleAbs(
    image,
    alpha=1,
    beta=brightness
)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(bright_image)
plt.title("Brightness +50")
plt.axis("off")

plt.show()
#Giảm độ sáng của ảnh
dark_image = cv2.convertScaleAbs(
    image,
    alpha=1,
    beta=-50
)

plt.figure(figsize=(6,6))
plt.imshow(dark_image)
plt.title("Brightness -50")
plt.axis("off")
plt.show()
```

## 2. Thay đổi độ tương phản

Độ tương phản được thay đổi bằng cách nhân mỗi điểm ảnh với hệ số α.

Công thức:

g(x,y)=αf(x,y)

α > 1: tăng tương phản

0 < α < 1: giảm tương phản

```python
#Tăng độ tương phản
high_contrast = cv2.convertScaleAbs(
    image,
    alpha=1.5,
    beta=0
)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(high_contrast)
plt.title("Contrast x1.5")
plt.axis("off")

plt.show()
#Giảm độ tương phản
low_contrast = cv2.convertScaleAbs(
    image,
    alpha=0.5,
    beta=0
)

plt.figure(figsize=(6,6))
plt.imshow(low_contrast)
plt.title("Contrast x0.5")
plt.axis("off")
plt.show()
```

## 3.Biến đổi âm bản

Thực hiện biến đổi âm bản (Negative Transformation) cho một hình ảnh bằng thư viện  Matplotlib trong Python với công thức

s = L - 1 - r

Trong đó:

r: Giá trị của điểm ảnh gốc.

s: Giá trị của điểm ảnh sau khi biến đổi (âm bản).

L: Số mức xám/màu của ảnh

Với ảnh 8-bit chuẩn (RGB), L = 256, do đó L - 1 = 255.

Sẽ cho ra kết quả 

Bên trái: Ảnh gốc hệ màu RGB.

Bên phải: Ảnh âm bản đã biến đổi.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# Đường dẫn đến file ảnh trong folder datatest
image_path = os.path.join('..', 'datatest', 'test.jpg')


# Đọc ảnh
img = cv2.imread(image_path)

if img is None:
    print(f"Không tìm thấy ảnh tại: {image_path}. Bạn hãy kiểm tra lại tên file ảnh trong folder 'datatest' nhé!")
else:
    # Chuyển từ BGR sang RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # 3. Biến đổi âm bản: Đảo ngược các giá trị điểm ảnh
    negative_img = 255 - img_rgb

    # Hiển thị kết quả
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title('Ảnh gốc (RGB)')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(negative_img)
    plt.title('Phần 3: Ảnh âm bản')
    plt.axis('off')
    plt.show()
```

# 4.Cắt ngưỡng

Thực hiện kỹ thuật Cắt ngưỡng nhị phân (Binary Thresholding) trên ảnh xám bằng thư viện OpenCV và hiển thị kết quả bằng Matplotlib với công thức:


Trong đó:f(x, y): Giá trị độ sáng của điểm ảnh gốc tại vị trí (x, y) trong gray_img.

g(x, y): Giá trị của điểm ảnh sau khi cắt ngưỡng trong binary_img.

255: Giá trị tối đa gán cho điểm ảnh vượt ngưỡng (tương ứng với màu Trắng).

0: Giá trị gán cho điểm ảnh nhỏ hơn hoặc bằng ngưỡng (tương ứng với màu Đen).

### II. Làm sắc nét

Làm sắc nét: Tăng cường các cạnh trong ảnh bằng cách sử dụng các kernel thích hợp.

1. Kernel cơ bản: Nhân giá trị điểm ảnh trung tâm với hệ số lớn, trừ đi các điểm ảnh lân cận (trên, dưới, trái, phải) để khuếch đại sự khác biệt tại các cạnh, giữ nguyên vùng ảnh phẳng.
2. Kernel mạnh hơn: Trừ cả 8 điểm ảnh lân cận (bao gồm đường chéo), giúp cạnh rõ nét hơn nhiều nhưng cũng dễ khuếch đại nhiễu nếu ảnh gốc không sạch.
3. Unsharp masking: Làm mờ ảnh gốc bằng Gaussian blur để tách phần chi tiết cao tần (ảnh gốc trừ ảnh mờ), sau đó cộng ngược phần chi tiết đó vào ảnh gốc với hệ số điều chỉnh được, cho kết quả tự nhiên hơn.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


def sharpen_basic(image):
    """Làm sắc nét ảnh bằng kernel cơ bản (dựa trên Laplacian).

    Args:
        image: Ảnh đầu vào dạng NumPy array.

    Returns:
        Ảnh đã được làm sắc nét, dạng NumPy array.
    """
    kernel = np.array([
        [ 0, -1,  0],
        [-1,  5, -1],
        [ 0, -1,  0]
    ])
    sharpened = cv2.filter2D(image, -1, kernel)
    return sharpened


# Đọc ảnh
image = cv2.imread(r'C:\Users\ASUS\computer_vision\datatest\test.jpg')

# Áp dụng làm sắc nét
sharpened_image = sharpen_basic(image)

# Hiển thị so sánh
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Ảnh gốc')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(sharpened_image, cv2.COLOR_BGR2RGB))
plt.title('Sau khi làm sắc nét')
plt.axis('off')
plt.show()
```

```python
def sharpen_strong(image):
    """Làm sắc nét ảnh bằng kernel mạnh hơn (tăng cường cạnh nhiều hơn).

    Args:
        image: Ảnh đầu vào dạng NumPy array.

    Returns:
        Ảnh đã được làm sắc nét, dạng NumPy array.
    """
    kernel = np.array([
        [-1, -1, -1],
        [-1,  9, -1],
        [-1, -1, -1]
    ])
    sharpened = cv2.filter2D(image, -1, kernel)
    return sharpened


def sharpen_unsharp_mask(image, sigma=3, strength=1.5):
    """Làm sắc nét ảnh bằng kỹ thuật unsharp masking.

    Nguyên lý: làm mờ ảnh bằng Gaussian blur để lấy phiên bản "mất chi tiết",
    sau đó lấy ảnh gốc trừ đi ảnh mờ để ra phần chi tiết/cạnh, rồi cộng
    ngược phần chi tiết đó vào ảnh gốc với hệ số strength.

    Args:
        image: Ảnh đầu vào dạng NumPy array.
        sigma: Độ lệch chuẩn của Gaussian blur (kiểm soát vùng ảnh hưởng).
        strength: Hệ số khuếch đại phần chi tiết được cộng thêm.

    Returns:
        Ảnh đã được làm sắc nét, dạng NumPy array.
    """
    blurred = cv2.GaussianBlur(image, (0, 0), sigma)
    sharpened = cv2.addWeighted(image, 1 + strength, blurred, -strength, 0)
    return sharpened


# Áp dụng thêm 2 kernel còn lại
strong_image = sharpen_strong(image)
unsharp_image = sharpen_unsharp_mask(image)

# Hiển thị so sánh cả 4 ảnh: gốc, cơ bản, mạnh, unsharp masking
plt.figure(figsize=(16, 8))

plt.subplot(1, 4, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Ảnh gốc')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(cv2.cvtColor(sharpened_image, cv2.COLOR_BGR2RGB))
plt.title('Kernel cơ bản')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(cv2.cvtColor(strong_image, cv2.COLOR_BGR2RGB))
plt.title('Kernel mạnh hơn')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(cv2.cvtColor(unsharp_image, cv2.COLOR_BGR2RGB))
plt.title('Unsharp masking')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## III.1 - Phát hiện cạnh: Sobel vs Prewitt
Trần Thị Minh Thùy-091306004639
III. Bài tập nâng cao - Mục 1: Phát hiện cạnh bằng kernel Sobel và Prewitt

Nội dung thực hiện:
1.Cài đặt thủ công phép tích chập 2D (convolution) bằng vòng lặp
2.Cài đặt phát hiện cạnh bằng toán tử Sobel (thủ công và dùng OpenCV `cv2.filter2D`)
3.Cài đặt phát hiện cạnh bằng toán tử Prewitt (thủ công và dùng OpenCV)
4.Ngưỡng hóa (thresholding) kết quả để thu được ảnh nhị phân biên cạnh
5.So sánh trực quan kết quả giữa cách tính thủ công và dùng thư viện OpenCV

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolve2d_manual(image, kernel):
    """
    Thực hiện phép toán tích chập (convolution) 2D thủ công bằng vòng lặp.
    """
    img_h, img_w = image.shape
    k_h, k_w = kernel.shape
    
    pad_h = k_h // 2
    pad_w = k_w // 2
    
    padded_image = np.zeros((img_h + 2 * pad_h, img_w + 2 * pad_w), dtype=np.float32)
    padded_image[pad_h:pad_h + img_h, pad_w:pad_w + img_w] = image
    
    output = np.zeros((img_h, img_w), dtype=np.float32)
    kernel_flipped = np.flip(np.flip(kernel, 0), 1)
    
    for y in range(img_h):
        for x in range(img_w):
            neighborhood = padded_image[y:y + k_h, x:x + k_w]
            output[y, x] = np.sum(neighborhood * kernel_flipped)
            
    return output

def edge_detection_sobel(image, manual=True):
    """
    Phát hiện cạnh bằng toán tử Sobel.
    """
    Kx = np.array([[-1, 0, 1],
                   [-2, 0, 2],
                   [-1, 0, 1]], dtype=np.float32)
    
    Ky = np.array([[-1, -2, -1],
                   [ 0,  0,  0],
                   [ 1,  2,  1]], dtype=np.float32)
    
    if manual:
        Gx = convolve2d_manual(image, Kx)
        Gy = convolve2d_manual(image, Ky)
    else:
        Gx = cv2.filter2D(image.astype(np.float32), -1, Kx)
        Gy = cv2.filter2D(image.astype(np.float32), -1, Ky)
        
    magnitude = np.sqrt(Gx**2 + Gy**2)
    magnitude = np.clip(magnitude, 0, 255).astype(np.uint8)
    
    return magnitude, Gx, Gy

def edge_detection_prewitt(image, manual=True):
    """
    Phát hiện cạnh bằng toán tử Prewitt.
    """
    Kx = np.array([[-1, 0, 1],
                   [-1, 0, 1],
                   [-1, 0, 1]], dtype=np.float32)
    
    Ky = np.array([[-1, -1, -1],
                   [ 0,  0,  0],
                   [ 1,  1,  1]], dtype=np.float32)
    
    if manual:
        Gx = convolve2d_manual(image, Kx)
        Gy = convolve2d_manual(image, Ky)
    else:
        Gx = cv2.filter2D(image.astype(np.float32), -1, Kx)
        Gy = cv2.filter2D(image.astype(np.float32), -1, Ky)
        
    magnitude = np.sqrt(Gx**2 + Gy**2)
    magnitude = np.clip(magnitude, 0, 255).astype(np.uint8)
    
    return magnitude, Gx, Gy

def apply_threshold(magnitude, threshold_val=100):
    """
    Ngưỡng hóa (Thresholding) để thu được ảnh nhị phân biên cạnh rõ nét.
    """
    binary_edges = np.zeros_like(magnitude)
    binary_edges[magnitude >= threshold_val] = 255
    return binary_edges
```

```python
# Đọc ảnh xám từ dataset có sẵn
img_path = '../datatest/test.jpg'  
img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError(f"Không đọc được ảnh tại path: {img_path}")

print(f"Đã đọc ảnh kích thước {img.shape}")

# Sobel
sobel_manual, sobel_Gx_m, sobel_Gy_m = edge_detection_sobel(img, manual=True)
sobel_opencv, _, _ = edge_detection_sobel(img, manual=False)

# Prewitt
prewitt_manual, prewitt_Gx_m, prewitt_Gy_m = edge_detection_prewitt(img, manual=True)
prewitt_opencv, _, _ = edge_detection_prewitt(img, manual=False)

# Ngưỡng hóa
thresh = 100
sobel_thresh = apply_threshold(sobel_manual, thresh)
prewitt_thresh = apply_threshold(prewitt_manual, thresh)
```

```python
fig, axes = plt.subplots(3, 3, figsize=(15, 15))

axes[0, 0].imshow(img, cmap='gray')
axes[0, 0].set_title("Ảnh gốc (Grayscale)")
axes[0, 0].axis('off')

im_gx = axes[0, 1].imshow(sobel_Gx_m, cmap='bwr')
axes[0, 1].set_title("Sobel Gx (Cạnh dọc)")
axes[0, 1].axis('off')
fig.colorbar(im_gx, ax=axes[0, 1], fraction=0.046, pad=0.04)

im_gy = axes[0, 2].imshow(sobel_Gy_m, cmap='bwr')
axes[0, 2].set_title("Sobel Gy (Cạnh ngang)")
axes[0, 2].axis('off')
fig.colorbar(im_gy, ax=axes[0, 2], fraction=0.046, pad=0.04)

axes[1, 0].imshow(sobel_manual, cmap='gray')
axes[1, 0].set_title("Sobel Độ lớn Gradient (Thủ công)")
axes[1, 0].axis('off')

axes[1, 1].imshow(sobel_opencv, cmap='gray')
axes[1, 1].set_title("Sobel Độ lớn Gradient (OpenCV)")
axes[1, 1].axis('off')

axes[1, 2].imshow(sobel_thresh, cmap='gray')
axes[1, 2].set_title(f"Sobel Nhị phân hóa (Ngưỡng={thresh})")
axes[1, 2].axis('off')

axes[2, 0].imshow(prewitt_manual, cmap='gray')
axes[2, 0].set_title("Prewitt Độ lớn Gradient (Thủ công)")
axes[2, 0].axis('off')

axes[2, 1].imshow(prewitt_opencv, cmap='gray')
axes[2, 1].set_title("Prewitt Độ lớn Gradient (OpenCV)")
axes[2, 1].axis('off')

axes[2, 2].imshow(prewitt_thresh, cmap='gray')
axes[2, 2].set_title(f"Prewitt Nhị phân hóa (Ngưỡng={thresh})")
axes[2, 2].axis('off')

plt.tight_layout()
plt.savefig("sobel_prewitt_result.png", dpi=150, bbox_inches='tight')
plt.show()
```

**Nhận xét:**
Kết quả tính thủ công (`convolve2d_manual`) và dùng thư viện OpenCV (`cv2.filter2D`) cho ra ảnh gradient gần như giống hệt nhau ở cả Sobel và Prewitt, chứng tỏ hàm tích chập tự viết đã cài đặt đúng thuật toán.
Sobel Gx bắt rõ các cạnh dọc/xiên như vành nón và các sợi tóc, trong khi Sobel Gy phản ứng yếu hơn nhiều vì ảnh gốc chứa ít cạnh ngang rõ nét.
So sánh Sobel và Prewitt: hai kết quả khá tương đồng, nhưng Sobel cho đường biên sắc nét và ít nhiễu hơn một chút ở các chi tiết nhỏ (vùng tóc), do trọng số tâm của kernel Sobel lớn hơn (hệ số 2) giúp làm mượt nhiễu tốt hơn Prewitt.
Sau khi ngưỡng hóa (threshold=100), các đường viền chính của khuôn mặt và vành nón được giữ lại rõ ràng, tuy nhiên vùng tóc xoăn vẫn còn khá nhiều nhiễu dạng đốm nhỏ, có thể cải thiện bằng cách tăng ngưỡng hoặc áp dụng lọc làm mượt (Gaussian blur) trước khi phát hiện cạnh.

```python
#Tự thiết kế kernel: Tạo các kernel tùy chỉnh để đạt được các hiệu ứng khác nhau.
import numpy as np
# 1. Dò cạnh theo chiều ngang (Sobel Horizontal) - Hiện rõ các đường nằm ngang
kernel_sobel_h = np.array([
        [-1, -2, -1],
        [ 0,  0,  0],
        [ 1,  2,  1]
    ], dtype=np.float32)

    # 2. Dò cạnh theo chiều dọc (Sobel Vertical) - Hiện rõ các đường thẳng đứng
kernel_sobel_v = np.array([
        [-1, 0, 1],
        [-2, 0, 2],
        [-1, 0, 1]
    ], dtype=np.float32)

    # 3. Kernel Laplacian (Dò cạnh mọi hướng - Toán tử vi phân bậc 2)
kernel_laplacian = np.array([
        [ 0,  1,  0],
        [ 1, -4,  1],
        [ 0,  1,  0]
    ], dtype=np.float32)

    # 4. Kernel Siêu Sắc Nét (Unsharp Masking / Excess Sharpen)
kernel_super_sharpen = np.array([
        [-1, -1, -1],
        [-1,  9, -1],
        [-1, -1, -1]
    ], dtype=np.float32)

    # 5. Kernel Dịch Chuyển Bóng / Tạo Hiệu Ứng Chuyển Động (Motion Blur ngang)
size = 9
kernel_motion = np.zeros((size, size), dtype=np.float32)
kernel_motion[int((size - 1) / 2), :] = 1.0 / size  # Hàng giữa toàn số 1/size

results = {
        "1. Canh Ngang (Sobel H)": cv2.filter2D(image, -1, kernel_sobel_h),
        "2. Canh Doc (Sobel V)": cv2.filter2D(image, -1, kernel_sobel_v),
        "3. Do Canh Moi Huong (Laplacian)": cv2.filter2D(image, -1, kernel_laplacian),
        "4. Sieu Sac Net (Super Sharpen)": cv2.filter2D(image, -1, kernel_super_sharpen),
        "5. Lam Mo Chuyen Dong (Motion Blur)": cv2.filter2D(image, -1, kernel_motion)
    }


width, height = 400, 300

    # Hiện ảnh gốc trước
cv2.namedWindow("Anh Goc", cv2.WINDOW_NORMAL)
cv2.resizeWindow("Anh Goc", width, height)
cv2.imshow("Anh Goc", image)

    # Loop qua danh sách kết quả và mở cửa sổ
for win_name, img_res in results.items():
        cv2.namedWindow(win_name, cv2.WINDOW_NORMAL)
        cv2.resizeWindow(win_name, width, height)
        cv2.imshow(win_name, img_res)

cv2.waitKey(0)
cv2.destroyAllWindows()          
```

```python
#So sánh các loại lọc: So sánh hiệu quả của các loại lọc khác nhau trên cùng một hình ảnh
import matplotlib.pyplot as plt

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    

    # Lọc 1: Average Blur (Làm mờ trung bình)
blur_avg = cv2.blur(image_rgb, (5, 5))

    # Lọc 2: Gaussian Blur (Làm mờ mịn Gauss)
blur_gaussian = cv2.GaussianBlur(image_rgb, (5, 5), 0)

    # Lọc 3: Median Blur (Làm mờ trung vị - Giảm nhiễu tốt)
blur_median = cv2.medianBlur(image_rgb, 5)

    # Lọc 4: Sobel Filter (Dò cạnh - Cần dùng ảnh xám)
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
sobel_x = cv2.Sobel(image_gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(image_gray, cv2.CV_64F, 0, 1, ksize=3)
sobel_combined = cv2.magnitude(sobel_x, sobel_y)

    # Lọc 5: Laplacian Filter (Dò viền đa hướng)
laplacian = cv2.Laplacian(image_gray, cv2.CV_64F)

titles = [
        "1. Ảnh Gốc (Đã chuẩn màu)",
        "2. Average Blur (cv2.blur)",
        "3. Gaussian Blur (cv2.GaussianBlur)",
        "4. Median Blur (cv2.medianBlur)",
        "5. Dò Cạnh Sobel (cv2.Sobel)",
        "6. Dò Viền Laplacian (cv2.Laplacian)"
    ]

images = [
        image_rgb, 
        blur_avg, 
        blur_gaussian, 
        blur_median, 
        sobel_combined, 
        laplacian
    ]

plt.figure(figsize=(15, 9))

for i in range(6):
        plt.subplot(2, 3, i + 1)
        
        if i >= 4:
            plt.imshow(images[i], cmap='gray')
        else:
            plt.imshow(images[i])
            
        plt.title(titles[i], fontsize=11, fontweight='bold', pad=8)
        plt.axis('off')

plt.tight_layout()
plt.show()
```

So sánh hiệu quả của các loại lọc khác nhau trên cùng một hình ảnh

**1.Nhóm bộ lọc Làm mờ & Khử nhiễu (Smoothing / Blurring)**

**Average Blur & Gaussian Blur**: Làm mịn các vùng ảnh, giảm thiểu nhiễu hạt (Noise)

**Median Blur**:Xử lý cực kỳ xuất sắc đối với các loại nhiễu dạng chấm (Salt-and-Pepper noise).

**Đánh giá**: Khác với Gaussian, Median Blur vẫn giữ lại độ sắc nét tương đối của các đường biên (Edges) mà không làm nhòe mảng màu xung quanh, rất thích hợp cho bước tiền xử lý ảnh.

******2.Nhóm bộ lọc Dò cạnh & Tách biên:**

**Sobel Filter**: Xác định ranh giới đối tượng bằng đạo hàm bậc 1 theo hướng ngang (X) và hướng dọc (Y).

**Đánh giá**: Cung cấp thông tin về hướng của độ biến thiên màu sắc. Tuy nhiên, đường viền thu được thường dầy và dễ bị ảnh hưởng bởi nhiễu nền.

**Laplacian Filter**:Sử dụng đạo hàm bậc 2 để dò viền đa hướng đồng thời.

**Đánh giá**: Bắt được các chi tiết viền rất nhỏ và mảnh, nhưng cực kỳ nhạy cảm với nhiễu (thường đòi hỏi phải lọc mờ bằng Gaussian trước khi áp dụng).
