Nội dung thực hiện: Toán tử điện ảnh phần.
- 3. Biến đổi âm bản (Negative Transformation)
- 4. Cắt ngưỡng (Thresholding)


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# Đường dẫn đến file ảnh trong folder datatest
# Bạn kiểm tra tên ảnh trong folder datatest rồi thay vào bên dưới nhé (ví dụ: 'test.jpg' hoặc 'lenna.png')
image_name = "test.jpg"
image_path = os.path.join("datatest", image_name)

img = cv2.imread(image_path)

if img is None:
    print(f"Không tìm thấy ảnh tại: {image_path}. Bạn hãy kiểm tra lại tên file ảnh trong folder 'datatest' nhé!")
else:
    # Chuyển từ BGR sang RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # 3. Biến đổi âm bản: Đảo ngược các giá trị điểm ảnh
    negative_img = 255 - img_rgb

    # Hiển thị kết quả
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title('Ảnh gốc (RGB)')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(negative_img)
    plt.title('Phần 3: Ảnh âm bản')
    plt.axis('off')
    plt.show()
```

```python
if img is not None:
    # Chuyển ảnh sang ảnh xám để cắt ngưỡng
    gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 4. Cắt ngưỡng: So sánh giá trị điểm ảnh với ngưỡng T (ví dụ T = 128)
    threshold_value = 128
    _, binary_img = cv2.threshold(gray_img, threshold_value, 255, cv2.THRESH_BINARY)

    # Hiển thị kết quả
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(gray_img, cmap='gray')
    plt.title('Ảnh xám gốc')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(binary_img, cmap='gray')
    plt.title(f'Phần 4: Cắt ngưỡng (Threshold = {threshold_value})')
    plt.axis('off')
    plt.show()
```
